# Template visualization in a notebook

`PipelineView` shows the template of a `CoSyLuigiRepo` interactively: drag nodes, pan and zoom, hover a node to highlight its edges. The repository classes come from `visualization_example.py`.

```python
from visualization_example import SubstituteNameTask, WriteTemplateTask, WriteTitleTemplateTask

from cosy_luigi import CoSyLuigiRepo
from cosy_luigi.visualization.interactive import PipelineView

repo = CoSyLuigiRepo(WriteTemplateTask, WriteTitleTemplateTask, SubstituteNameTask)
PipelineView(repo, height=450)
```

A `{'taxonomy': ..., 'requires': ...}` dict works as well (`requires` is keyed by the class declaring the parameter). `save(...)` writes the same page as a standalone HTML file, e.g. for a review; `inline_elk=True` embeds elkjs so that the file also works offline.

```python
info = {
    "taxonomy": {
        "GetCosts": ["CoSyLuigiTask"],
        "GetHistoricDemand": ["CoSyLuigiTask"],
        "PredictDemandByAverage": ["CoSyLuigiTask", "PredictDemand"],
        "PredictDemandByLinearRegression": ["CoSyLuigiTask", "PredictDemand"],
        "OptimizeLotsByGroff": ["CoSyLuigiTask", "OptimizeLots"],
        "OptimizeLotsBySilverMeal": ["CoSyLuigiTask", "OptimizeLots"],
    },
    "requires": {"PredictDemand": ["GetHistoricDemand"], "OptimizeLots": ["GetCosts", "PredictDemand"]},
}

view = PipelineView(info, height=350, title="Lot sizing")
view.save("output/lot_sizing_review.html")
view
```
